```python
mis_num = np.array([10,20,np.nan,40,np.nan,50])
print(mis_num)
```

```python
mis_num[~np.isnan(mis_num)] #--> to get non_missing values we can use boolen mask
```

```python
# to calculate sum, mean, median , std, var while ignoring missing values we use nansum, nanmean, nanmedian, nanstd, nanvar

print(np.nansum(mis_num))
print(np.nanmean(mis_num))
print(np.nanmedian(mis_num))
print(np.nanstd(mis_num))
print(np.nanvar(mis_num))
```
